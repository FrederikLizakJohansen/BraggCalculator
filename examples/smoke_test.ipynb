```python
from braggcalculator import BraggCalculator
from braggcalculator.backends.numpy_backend import NumpyBackend
import matplotlib.pyplot as plt

calc = BraggCalculator(
    mode="xray",
    wavelength=1.5406,
    backend=NumpyBackend(),
)
calc.load("NaCl.cif")
tt, I = calc.pattern()

print(tt.shape, I.shape)       # sanity
print(calc._symm["spacegroup_symbol"])  # e.g., Fm-3m

plt.plot(tt, I)
plt.show()
```
